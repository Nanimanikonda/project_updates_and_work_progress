```python
#used to drop duplicate records. if we use inplace=True the changes permanently applies to data else just for understanding
df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

```python
#selecting rows based on condition
df['Date'] = df['Date'].astype('string').str.strip("'")
df['Date'] = pd.to_datetime(df['Date'], format='mixed', errors='coerce')
pd1=df[(df['Date'].between('2020/12/12','2020/12/23'))]
print(pd1)
```

```python
pd1.fillna(0,inplace=True)
print(pd1)
```
